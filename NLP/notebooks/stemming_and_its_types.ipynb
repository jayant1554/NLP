```python
from nltk.stem import PorterStemmer
```

```python
portstem=PorterStemmer()
```

```python
words=["run","runner","running","ran","runs","easily","fairly","fairness",'go','going','goes','gone','eat','eating','eats','ate']
```

```python
for word in words:
    print(word, " : ", portstem.stem(word))
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowstem=SnowballStemmer("english")
```

```python
for word in words:
    print(word, " : ", snowstem.stem(word))
```

```python
snowstem.stem("fairly"), snowstem.stem("fairness")
```
