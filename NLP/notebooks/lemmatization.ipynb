```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizor=WordNetLemmatizer()
```

```python
"""Pos
noun=n default
verb=v
adjective=a
adverb=r
"""
```

```python
import nltk

nltk.download("wordnet", quiet=True)
lemmatizor.lemmatize("goes")
```

```python
lemmatizor.lemmatize("goes", pos="a")
```

```python
lemmatizor.lemmatize("fairly", pos="r"), lemmatizor.lemmatize("fairness", pos="n")
```

```python
words=["run","runner","running","ran","runs","easily","fairly","fairness",'go','going','goes','gone','eat','eating','eats','ate']
```

```python
for word in words:
    print(word, " : ", lemmatizor.lemmatize(word, pos="v"))
```

```python
for word in words:
    print(word, " : ", lemmatizor.lemmatize(word, pos="n"))
```

```python
for word in words:
    print(word, " : ", lemmatizor.lemmatize(word, pos="a"))
```

```python
for word in words:
    print(word, " : ", lemmatizor.lemmatize(word, pos="r"))
```
