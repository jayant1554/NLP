### STEMMING AND ITS TYPE


```python
#used in 
#classification prob
# cooments of product is positive or negative
# sentiment analysis
# text classification
```

```python
words = ['running', 'ran', 'eating', 'eats', 'eaten', 'eater','teach, teaches', 'teaching', 'taught', 'teachings','walk', 'walks', 'walking', 'walked', 'walker']
```

### porterStemmer

```python
from nltk.stem import PorterStemmer
```

```python
ps=PorterStemmer()
```

```python
for word in words:
    print(word, " : ", ps.stem(word))
```

```python
ps.stem("congratulation")
#disadv
```

```python
#REGEX STEMMING
#regular expression
```

```python
from nltk.stem import RegexpStemmer
```

```python
regex= RegexpStemmer('ing$|s$|es$|ed$', min=4)
```

```python
for word in words:
    print(word, " :  " , regex.stem(word))
```

```python
regex.stem('playing')
```

```python
regex.stem("ingplaying") #didn't work in prefix
```

## snowball stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snowstemmer= SnowballStemmer('english')
```

```python
for word in words:
    print(word,"  :  ",snowstemmer.stem(word))
```

```python
snowstemmer.stem("goes")
```
